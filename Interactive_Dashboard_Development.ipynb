```python

import pandas as pd

data = {
    'Order_ID': [101,102,103,104,105,106,107,108,109,110,
                 111,112,113,114,115,116,117,118,119,120],
    'Product': ['Laptop','Mobile','Tablet','Laptop','Headphones',
                'Mobile','Tablet','Laptop','Mobile','Headphones',
                'Laptop','Tablet','Mobile','Laptop','Headphones',
                'Mobile','Tablet','Laptop','Mobile','Headphones'],
    'Category': ['Electronics','Electronics','Electronics','Electronics',
                 'Accessories','Electronics','Electronics','Electronics',
                 'Electronics','Accessories','Electronics','Electronics',
                 'Electronics','Electronics','Accessories','Electronics',
                 'Electronics','Electronics','Electronics','Accessories'],
    'Quantity': [2,3,1,1,4,2,3,2,1,5,1,2,3,2,3,1,4,1,2,4],
    'Price': [50000,20000,15000,50000,2000,20000,15000,50000,20000,2000,
              50000,15000,20000,50000,2000,20000,15000,50000,20000,2000],
    'Region': ['South','North','East','West','South','North','East','West',
               'South','North','East','West','South','North','East',
               'West','South','North','East','West']
}

df = pd.DataFrame(data)

df['Sales'] = df['Quantity'] * df['Price']

print("Dataset created successfully!")
```

```python

df
```

```python

print("Dataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe())
```

```python

print("Missing Values:")
print(df.isnull().sum())
```

```python

# Check duplicate records
print("Duplicate Records:", df.duplicated().sum())

# Remove duplicates
df = df.drop_duplicates()

# Handle missing values
df = df.dropna()

print("Data Cleaning Completed!")
print("Final Dataset Shape:", df.shape)
```

```python

# Total Sales
total_sales = df['Sales'].sum()

# Total Orders
total_orders = df['Order_ID'].nunique()

# Total Customers (unique regions as a sample proxy)
total_customers = df['Region'].nunique()

# Average Sales
average_sales = df['Sales'].mean()

# Total Quantity Sold
total_quantity = df['Quantity'].sum()

print("Total Sales:", total_sales)
print("Total Orders:", total_orders)
print("Total Regions:", total_customers)
print("Average Sales:", round(average_sales, 2))
print("Total Quantity Sold:", total_quantity)
```

```python

product_sales = df.groupby('Product')['Sales'].sum()

print("Sales by Product:")
display(product_sales.sort_values(ascending=False))

print("Best Selling Product:", product_sales.idxmax())
```

```python

region_sales = df.groupby('Region')['Sales'].sum()

print("Sales by Region:")
display(region_sales.sort_values(ascending=False))

print("Highest Sales Region:", region_sales.idxmax())
```

```python

category_sales = df.groupby('Category')['Sales'].sum()

print("Sales by Category:")
display(category_sales)
```

```python

import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

product_sales.sort_values().plot(
    kind='bar',
    title='Sales by Product'
)

plt.xlabel('Product')
plt.ylabel('Total Sales')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python

plt.figure(figsize=(8,5))

region_sales.plot(kind='bar', title='Sales by Region')

plt.xlabel('Region')
plt.ylabel('Total Sales')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python

plt.figure(figsize=(7,7))

plt.pie(
    category_sales,
    labels=category_sales.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Sales Distribution by Category')
plt.show()
```

```python

quantity_sales = df.groupby('Product')['Quantity'].sum()

plt.figure(figsize=(8,5))

quantity_sales.plot(
    kind='bar',
    title='Quantity Sold by Product'
)

plt.xlabel('Product')
plt.ylabel('Total Quantity')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python

%%writefile sales_dashboard.py

import streamlit as st
import pandas as pd
import plotly.express as px

# Page configuration
st.set_page_config(
    page_title="Sales Analytics Dashboard",
    page_icon="📊",
    layout="wide"
)

# Load dataset
df = pd.read_csv("sales_data.csv")

# Dashboard title
st.title("📊 Sales Analytics Dashboard")
st.write("Interactive Business Sales Performance Analysis")

# Sidebar filters
st.sidebar.header("Dashboard Filters")

products = st.sidebar.multiselect(
    "Select Products",
    options=df["Product"].unique(),
    default=df["Product"].unique()
)

regions = st.sidebar.multiselect(
    "Select Regions",
    options=df["Region"].unique(),
    default=df["Region"].unique()
)

categories = st.sidebar.multiselect(
    "Select Categories",
    options=df["Category"].unique(),
    default=df["Category"].unique()
)

# Filter dataset
filtered_df = df[
    (df["Product"].isin(products)) &
    (df["Region"].isin(regions)) &
    (df["Category"].isin(categories))
]

st.subheader("📌 Key Performance Indicators")

if not filtered_df.empty:

    total_sales = filtered_df["Sales"].sum()
    total_orders = filtered_df["Order_ID"].nunique()
    total_quantity = filtered_df["Quantity"].sum()
    average_sales = filtered_df["Sales"].mean()

    col1, col2, col3, col4 = st.columns(4)

    col1.metric("Total Sales", f"₹{total_sales:,.0f}")
    col2.metric("Total Orders", total_orders)
    col3.metric("Quantity Sold", total_quantity)
    col4.metric("Average Sales", f"₹{average_sales:,.0f}")

    st.divider()

    # Sales by Product
    st.subheader("Sales by Product")

    product_sales = (
        filtered_df.groupby("Product", as_index=False)["Sales"].sum()
    )

    fig1 = px.bar(
        product_sales,
        x="Product",
        y="Sales",
        color="Product",
        title="Product Sales Performance"
    )

    st.plotly_chart(fig1, use_container_width=True)

    # Sales by Region
    col5, col6 = st.columns(2)

    with col5:
        region_sales = (
            filtered_df.groupby("Region", as_index=False)["Sales"].sum()
        )

        fig2 = px.pie(
            region_sales,
            names="Region",
            values="Sales",
            title="Sales Distribution by Region"
        )

        st.plotly_chart(fig2, use_container_width=True)

    # Sales by Category
    with col6:
        category_sales = (
            filtered_df.groupby("Category", as_index=False)["Sales"].sum()
        )

        fig3 = px.bar(
            category_sales,
            x="Category",
            y="Sales",
            color="Category",
            title="Sales by Category"
        )

        st.plotly_chart(fig3, use_container_width=True)

    # Sales table
    st.subheader("📋 Sales Data")
    st.dataframe(filtered_df, use_container_width=True)

    # Business insights
    st.subheader("💡 Business Insights")

    best_product = product_sales.loc[
        product_sales["Sales"].idxmax(), "Product"
    ]

    best_region = region_sales.loc[
        region_sales["Sales"].idxmax(), "Region"
    ]

    st.write("🏆 Best Performing Product:", best_product)
    st.write("📍 Highest Sales Region:", best_region)
    st.write("💰 Total Revenue: ₹", f"{total_sales:,.0f}")

else:
    st.warning("No data available. Please select at least one filter.")

st.success("Dashboard loaded successfully!")
```
